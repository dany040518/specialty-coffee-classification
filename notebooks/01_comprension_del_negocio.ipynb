# 01 — Comprensión del negocio (Business Understanding)

## 1.1 Contexto y problema de negocio

Detrás de cada taza de café hay decisiones que empiezan mucho antes de la cosecha: la tierra, la variedad, la altura de la finca, cómo se procesa el grano. Para Colombia eso no es un detalle menor — el café sostiene a cientos de miles de familias caficultoras, en su mayoría pequeños productores, y es lo que convirtió al país en el tercer productor mundial (detrás de Brasil y Vietnam) y el principal productor de café arábico suave lavado [1][3]. Cualquier factor que ayude a que un lote alcance mejor calidad se traduce en mejores ingresos para quien lo cultiva.

¿Qué hace que un café sea "de calidad"? La industria no lo deja al azar: el protocolo más reconocido es el de la *Specialty Coffee Association (SCA)*, donde un panel de catadores certificados evalúa aroma, sabor, acidez, cuerpo y balance, y construye un puntaje sobre 100. Por encima de 80, el lote entra a la categoría de "café de especialidad" [5], una distinción que suele venir con mejor precio para el productor.

El problema es que ese puntaje solo se conoce al final, cuando el café ya fue cosechado, trillado y catado — para entonces lo que más influyó en el resultado ya quedó atrás. De ahí la pregunta que guía este proyecto: en vez de esperar hasta la taza final e imponer un único umbral de calidad, ¿pueden los datos revelar por sí solos si existen distintos perfiles de café y qué los caracteriza? Ese entendimiento puede guiar a un productor sobre qué renovar, qué variedad sembrar o cómo estandarizar el procesamiento.

## 1.2 Objetivos

### Objetivo general

Explorar, mediante clustering, si los lotes de café se agrupan en distintos perfiles de calidad, y qué características de cultivo y manejo (origen, variedad, procesamiento) se asocian con los que mejor califican — como guía práctica para decidir qué sembrar o cómo estandarizar el manejo poscosecha.

### Objetivos específicos

- Encontrar grupos naturales a partir de los atributos en taza (aroma, sabor, acidez, cuerpo, balance), probando distintas formas de agruparlos hasta quedarnos con la que mejor refleje esas diferencias.
- Revisar qué tienen en común los lotes de cada grupo en origen y manejo (país, altura, variedad, procesamiento).
- Mostrar los grupos y lo que los caracteriza en un dashboard fácil de entender, pensado para el sector cafetero.

## 1.3 Evaluación de la situación

### Recursos

Se usa una base pública del Coffee Quality Institute (CQI), la misma organización detrás del protocolo SCA — fuente reconocida, usada en proyectos similares, con cataciones de arábica y robusta de varios países (Colombia incluida). Ya está en el repositorio, en `data/raw/`: `arabica_data_cleaned.csv` y `robusta_data_cleaned.csv`.

El trabajo se hace en Python: pandas para los datos, scikit-learn para el clustering, matplotlib/seaborn para visualizar. Los resultados se comunican en un dashboard en Power BI o Tableau.

### Supuestos

Que las cataciones siguieron el protocolo SCA de forma consistente entre lotes (puntajes comparables entre sí), y que la información de origen y manejo registrada corresponde al lote que fue catado.

### Restricciones

Se trabaja solo con los datos ya disponibles públicamente, sin recolectar información adicional. El proyecto tiene un plazo de dos semanas, lo que exige decisiones ágiles y bien justificadas. Arábica y robusta se evalúan con formularios de catación distintos, así que probablemente haya que tratarlas por separado.

### Riesgos

Algunos países o variedades pueden tener muy pocos registros, dificultando sacar conclusiones firmes sobre ellos. La información de origen y manejo puede estar incompleta o poco estandarizada, según cómo la reportó cada productor o exportador. Se confirman (y se tratan, si hace falta) en las siguientes fases.

## 1.4 Objetivos de minería de datos

En términos técnicos:

| Aspecto | Descripción |
|---|---|
| Tipo de problema | Aprendizaje no supervisado, mediante clustering: no hay etiqueta de calidad predefinida, se busca que los datos revelen agrupaciones naturales. |
| Unidad de análisis | Cada lote de café evaluado (cada fila de la base). |
| Variables para agrupar | Los atributos de catación del protocolo SCA: aroma, sabor, acidez, cuerpo, balance, entre otros. |
| Variables para caracterizar los grupos | Condiciones de origen y manejo: país, altitud, variedad, método de procesamiento. |
| Resultados esperados | Un número de grupos definidos y justificados, qué distingue a cada uno, y los insumos para el dashboard final. |

## 1.5 Criterios de éxito

### Criterio de éxito de negocio

El proyecto es exitoso si señala, con respaldo cuantitativo, qué condiciones de origen y manejo se asocian con los grupos de mejor calidad — lo bastante claro como para guiar decisiones del sector — y si el dashboard comunica esos hallazgos a una audiencia no técnica.

### Criterio de éxito de minería de datos

Los grupos deben ser claramente diferenciables e interpretables, evaluados con medidas de calidad de agrupamiento (como el coeficiente de silueta), y las características que distinguen a cada grupo deben ser consistentes con lo que se observe en la exploración de los datos.

Los valores exactos de estas medidas se terminan de fijar más adelante, cuando ya se conozca el comportamiento real de los datos.

## 1.6 Plan del proyecto

| Fase | Qué incluye |
|---|---|
| 1. Comprensión del negocio | Plantea el problema, los objetivos, la evaluación de la situación y los criterios de éxito. |
| 2. Comprensión de los datos | Explora la base de datos, revisa su calidad y deja los primeros hallazgos que orientan la preparación. |
| 3. Preparación de los datos | Limpia y transforma la información, y deja lista la tabla final que se usará para el clustering. |
| 4. Modelado | Aplica y compara distintas técnicas de clustering, y selecciona la que mejor representa los datos. |
| 5. Evaluación | Revisa qué tan buenos y consistentes son los grupos obtenidos, contrastándolos con los criterios de éxito. |
| 6. Interpretación y resultados | Analiza qué caracteriza a cada grupo y traduce esos hallazgos en conclusiones para el sector cafetero, además de dejar lista la información para el dashboard. |
