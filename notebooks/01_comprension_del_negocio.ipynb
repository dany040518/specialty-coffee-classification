## Contexto y problema de negocio

La [Coffee Quality Institute (CQI) database](https://github.com/jldbc/coffee-quality-database) contiene más de 1300 registros de catación de lotes de café (arabica y robusta), evaluados profesionalmente bajo el protocolo de la Specialty Coffee Association (SCA).

A partir de estos datos se busca **predecir si un lote de café califica como "café de especialidad"** y **cuáles son las condiciones asociadas a esa clasificación**. El café es un producto de gran relevancia económica y social en Colombia, por lo que este análisis podría usarse para orientar al sector cafetero colombiano sobre qué factores (origen, altitud, variedad, método de procesamiento, etc.) se asocian con lotes de mayor calidad.
