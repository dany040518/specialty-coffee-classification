# 01 — Comprensión del negocio (Business Understanding)

## 1.1 Contexto y problema de negocio

Detrás de cada taza de café hay una cadena de decisiones que empieza mucho antes de la cosecha: la tierra donde se siembra, la variedad que se elige, la altura de la finca, la forma en que se procesa el grano, entre muchas otras. Para Colombia esa cadena no es un detalle menor: el café es una de las actividades más importantes del campo colombiano y sostiene a cientos de miles de familias caficultoras, en su mayoría pequeños productores. Es, además, lo que ha llevado al país a convertirse en el tercer productor mundial —solo detrás de Brasil y Vietnam— y en el principal productor de café arábico suave lavado [1][3]. Por eso, cualquier factor que ayude a que un lote alcance mejor calidad termina traduciéndose, tarde o temprano, en mejores ingresos para quienes lo cultivan, además de contribuir a mantener y afianzar el renombre que tiene el café colombiano en el mundo.

Pero, ¿qué hace que un café sea "de calidad"? La industria no lo deja al azar: existen protocolos de catación estandarizados, siendo el más reconocido en el mundo el de la *Specialty Coffee Association (SCA)*. Un panel de catadores certificados evalúa atributos como el aroma, el sabor, la acidez, el cuerpo y el balance de la taza, y con ellos construye un puntaje sobre 100 puntos. Cuando ese puntaje supera los 80, el lote entra a la categoría de "café de especialidad" [5], una distinción que no solo reconoce la calidad del café, sino que suele venir acompañada de beneficios adicionales, como un mejor precio para el productor.

El problema es que ese puntaje solo se conoce al final, cuando el café ya fue cosechado, trillado y catado. Para entonces, todo lo que más influyó en el resultado ya quedó atrás y no se puede cambiar. De ahí nace la pregunta que guía este proyecto: en lugar de esperar hasta la taza final, ¿es posible encontrar patrones o grupos naturales entre los lotes de café que ayuden a entender, de antemano, qué combinaciones tienden a producir mejores resultados? En vez de imponer desde el inicio un único umbral de calidad, buscamos dejar que los propios datos revelen si existen distintos perfiles de café y qué los caracteriza. Ese entendimiento puede convertirse en una guía práctica para productores a la hora de decidir qué renovar, qué variedad sembrar o cómo estandarizar el procesamiento del grano.

## 1.2 Objetivos

### Objetivo general

Explorar, mediante clustering, si los lotes de café pueden agruparse en distintos perfiles de calidad, y a partir de esos grupos entender qué características de su cultivo y su manejo, como el origen, la variedad o la forma de procesamiento, están más asociadas con los lotes que mejor califican. Buscamos que este análisis sirva como una guía para el sector cafetero a la hora de tomar decisiones, por ejemplo sobre qué variedades sembrar o cómo estandarizar el manejo poscosecha, con el fin de mejorar la calidad de sus cosechas.

### Objetivos específicos

- Encontrar, mediante clustering, grupos naturales entre los lotes de café a partir de sus atributos en la taza, como el aroma, el sabor, la acidez, el cuerpo y el balance, probando distintas formas de agruparlos hasta quedarnos con la que mejor refleje esas diferencias.
- Revisar qué tienen en común los lotes dentro de cada grupo en cuanto a su origen y su manejo, como el país, la altura, la variedad y la forma de procesamiento, para entender qué distingue a los que mejor califican de los demás.
- Mostrar los grupos encontrados y lo que los caracteriza en un dashboard fácil de entender, pensado para el sector cafetero.

## 1.3 Evaluación de la situación

Antes de empezar a trabajar con los datos, vale la pena dejar claro con qué se cuenta y qué se espera encontrar, para que las decisiones que vengan después tengan un punto de referencia.

### Recursos

Para este proyecto se usará una base de datos pública construida por el Coffee Quality Institute (CQI), la misma organización detrás del protocolo de catación SCA que se menciono anteriormente. Es una fuente reconocida y usada en proyectos similares, reúne cataciones de lotes de café arábica y robusta de distintos países, entre ellos Colombia, y ya está disponible dentro del repositorio del proyecto, en la carpeta `data/raw/`, donde se encuentran los dos conjuntos de datos: `arabica_data_cleaned.csv` y `robusta_data_cleaned.csv`.

El trabajo se hará en Python, apoyándonos en herramientas como pandas para organizar los datos, scikit-learn para el clustering y librerías de visualización como matplotlib o seaborn; los resultados se comunicarán al final en un dashboard construido en Power BI o Tableau.

### Supuestos

Como el trabajo con los datos todavía no ha comenzado, hay algunos supuestos que conviene dejar por escrito desde ya. Se asume que las cataciones se hicieron siguiendo el protocolo SCA de forma consistente entre lotes, de manera que los puntajes sean comparables entre sí, y que la información de origen y manejo registrada corresponde realmente al lote que fue catado.

### Restricciones

También hay restricciones propias del proyecto que van a condicionar lo que se puede hacer: se trabaja únicamente con los datos que ya están disponibles públicamente, sin posibilidad de recolectar información adicional; el proyecto tiene un plazo de dos semanas, lo que exige avanzar con decisiones ágiles y bien justificadas; y como arábica y robusta se evalúan con formularios de catación distintos, es probable que haya que tratarlas por separado o decidir con cuidado cómo combinarlas.

### Riesgos

Por último, hay algunos riesgos que se pueden anticipar, aunque todavía no están confirmados porque no hemos explorado los datos en detalle: es posible que algunos países o variedades tengan muy pocos registros, lo que haría difícil sacar conclusiones firmes sobre ellos, y es posible también que la información de origen y manejo esté incompleta o poco estandarizada, ya que suele depender de cómo cada productor o exportador la reportó. Todos estos riesgos se confirmarán y, de ser necesario, se tratarán en las siguientes fases.

## 1.4 Objetivos de minería de datos

Aterrizando el objetivo de negocio en términos más técnicos, así quedaría planteado nuestro análisis:

| Aspecto | Descripción |
|---|---|
| Tipo de problema | Aprendizaje no supervisado, mediante clustering: no se parte de una etiqueta de calidad predefinida, sino que se busca que los propios datos revelen agrupaciones naturales entre los lotes de café. |
| Unidad de análisis | Cada lote de café evaluado, es decir, cada fila de la base de datos. |
| Variables para agrupar | Los atributos de catación de cada lote, como el aroma, el sabor, la acidez, el cuerpo y el balance, entre otros que forman parte del protocolo SCA. |
| Variables para caracterizar los grupos | Las condiciones de origen y manejo de cada lote, como el país, la altitud, la variedad y el método de procesamiento. |
| Resultados esperados | Un número de grupos claramente definidos y justificados, una descripción de qué distingue a cada uno, y los insumos necesarios para construir un dashboard final. |

## 1.5 Criterios de éxito

Antes de entrar a los datos, conviene dejar claro qué se va a considerar un resultado exitoso, aunque los valores exactos se terminen de ajustar más adelante, cuando ya se conozca el comportamiento real de los datos.

### Criterio de éxito de negocio

El proyecto es exitoso si logra señalar, con respaldo cuantitativo, qué condiciones de origen y manejo se asocian con los grupos de mejor calidad, de forma lo suficientemente clara como para servir de guía a decisiones del sector cafetero, y si el dashboard logra comunicar esos hallazgos a una audiencia no técnica.

### Criterio de éxito de minería de datos

Los grupos obtenidos deben ser claramente diferenciables entre sí y fáciles de interpretar, evaluados con medidas de calidad de agrupamiento, como el coeficiente de silueta, y las características que distinguen a cada grupo deben ser consistentes con lo que se observe más adelante en la exploración de los datos.

## 1.6 Plan del proyecto

| Fase | Qué incluye |
|---|---|
| 1. Comprensión del negocio | Plantea el problema, los objetivos, la evaluación de la situación y los criterios de éxito. |
| 2. Comprensión de los datos | Explora la base de datos, revisa su calidad y deja los primeros hallazgos que orientan la preparación. |
| 3. Preparación de los datos | Limpia y transforma la información, y deja lista la tabla final que se usará para el clustering. |
| 4. Modelado | Aplica y compara distintas técnicas de clustering, y selecciona la que mejor representa los datos. |
| 5. Evaluación | Revisa qué tan buenos y consistentes son los grupos obtenidos, contrastándolos con los criterios de éxito. |
| 6. Interpretación y resultados | Analiza qué caracteriza a cada grupo y traduce esos hallazgos en conclusiones para el sector cafetero, además de dejar lista la información para el dashboard. |
